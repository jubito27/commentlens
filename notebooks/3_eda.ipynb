```python
import numpy as n
def sd(data):
    data = n.array(data)
    s = n.std(data)
    return s

def variance(data):
    data = n.array(data)
    v = n.var(data)
    return v

def covar(x , y):
    cv = n.cov(x , y)
    return cv

data = [3,2,4,1,5,2,5,3,2,5,3,5,3,5]
x= [2,3,45,2,4]
y = [2,6,4,56,6]
s = sd(data)
v = variance(data=data)
cv = covar(x , y)

print(f"Standard Deviation is {n.round(s , 2)}")
print(f"Variance is {n.round(v , 2)}")
print(f"COVariance is {n.round(cv , 2)}")


```
